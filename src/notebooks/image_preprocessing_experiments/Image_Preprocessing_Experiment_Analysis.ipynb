# Analysis of Image Preprocessing Experiments Results

```python
# Import libraries
import pandas as pd
import numpy as np
```

```python
# Load the results csv in a dataframe
results_df = pd.read_csv('image_preprocessing_results.csv')
```

## Custom CNN

```python
# Custom CNN all results
custom_cnn = results_df[results_df['model'] == 'Custom CNN']
```

```python
# Get Image preprocessing pipelines with max performances
max_auc_custom_cnn = custom_cnn[custom_cnn['auc'] == custom_cnn['auc'].max()]['image_preprocessing_pipeline']
max_sencitivity_custom_cnn = custom_cnn[custom_cnn['sensitivity'] == custom_cnn['sensitivity'].max()]['image_preprocessing_pipeline']
max_specificity_custom_cnn = custom_cnn[custom_cnn['specificity'] == custom_cnn['specificity'].max()]['image_preprocessing_pipeline']
max_precision_custom_cnn = custom_cnn[custom_cnn['precision'] == custom_cnn['precision'].max()]['image_preprocessing_pipeline']
max_f1_custom_cnn = custom_cnn[custom_cnn['f1_score'] == custom_cnn['f1_score'].max()]['image_preprocessing_pipeline']

# Get Image Preprocessing Pipelines with least computational costs
min_training_time_custom_cnn = custom_cnn[custom_cnn['training_time_mins'] == custom_cnn['training_time_mins'].min()]['image_preprocessing_pipeline']
min_prediction_time_custom_cnn = custom_cnn[custom_cnn['prediction_time_sec'] == custom_cnn['prediction_time_sec'].min()]['image_preprocessing_pipeline']
min_peak_memory_custom_cnn = custom_cnn[custom_cnn['peak_memory_used_MB'] == custom_cnn['peak_memory_used_MB'].min()]['image_preprocessing_pipeline']
```

```python
print(f"Max AUC:")
print(max_auc_custom_cnn)

print(f"Max Sensitivity:")
print(max_sencitivity_custom_cnn)

print(f"Max Specificity:")
print(max_specificity_custom_cnn)

print(f"Max Precision:")
print(max_precision_custom_cnn)

print(f"Max F1 Score:")
print(max_f1_custom_cnn)

print(f"Min Training Time (minutes):")
print(min_training_time_custom_cnn)

print(f"Min Prediction Time (seconds):")
print(min_prediction_time_custom_cnn)

print(f"Min Peak Memory (MB):")
print(min_peak_memory_custom_cnn)
```

### Inference for Custom CNN

* Max AUC: P4, P6
* Max Sensitivity: P2
* Max Precision: P4, P6
* Max F1-Score: P2
* Max Specificity: P3, P5, P7, P9, P10
* Min Training-Time: P4
* Min Prediction-Time: P4
* Min Peak Memory used: P9

Overall,
1. P2 attains max sensitivity and F1 Score.
2. P4 attains max AUC, Precision with minimum training and prediction time
3. P9 attains max specificity with minimum peak memory usage.

```python
custom_cnn[custom_cnn['image_preprocessing_pipeline'].isin(['P2', 'P4', 'P9'])]
```

### Result for Custom CNN
P2 and P4 have negligible difference in accuracy, AUC, precision, prediction time and peak memory usage values. However, P2 achieves significantly higher sensitivity, f1 score and training time (4 minutes more than P4).  

P9 achieves significantly lower sensitivity, precision and f1 scores with higher training and prediction times. Therefore, P9 can be eliminated.

Regardless of slightly higher training time, P2 demonstrates overall better performance across all metrics.

**Best Image Preprocessing Pipeline for Custom CNN: P2 (Resizing + CLAHE with 1-channel Grayscale + Normalization)**
